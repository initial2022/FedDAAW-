```python
"""
FedDAAW 消融实验 (MNIST, Non-IID) —— 三种参与预算一次跑完

4 组配置（逐组件消融，不再做 2x2 析因）：
  1. FedDAAW (Full) —— 完整版（DS + AA）
  2. w/o DS         —— 去掉动态客户端选择（随机选 + AA）
  3. w/o AA         —— 去掉自适应聚合权重（DS + 简单平均）
  4. FedAvg         —— 全拆基线（随机选 + 简单平均）

参与预算（每轮从 20 个客户端中选择）：
  k = 15（本轮只跑这一档；如需恢复三档改回 [15, 10, 5]）

三个种子：42 / 123 / 2024，每种组合训练 40 轮。

说明：
  - Client 与 ServerDA 完全保留原 notebook 实现，未做任何修改
  - 3 个消融变体通过子类化 ServerDA 实现，只覆盖需要改的方法
  - data_similarity 是原代码里的辅助信号，不单独消融
  - 同一 seed 下各预算共用同一份 Non-IID 划分，差异只来自参与预算
  - 逐变体早停（PATIENCE_ROUNDS = 3）：每个变体只看自己的历史最佳
    （best-so-far），连续 PATIENCE_ROUNDS 轮没有刷新就只停这一个，其余变体
    继续训练；因此各变体的 stop_round 与曲线长度可以不同。
    设 PATIENCE_ROUNDS = 0 则关闭早停，四个变体都跑满 NUM_GLOBAL_EPOCHS
  - 严格公平（已固定开启）：四个变体共享同一初始权重与同一组 sys_score，
    差异只来自选择/聚合策略；客户端划分也共用同一份 client_data

输出（results/ 目录，文件名带客户端数量；文件名带客户端数量）：
  ablation_mnist_result_k15.csv / k10 / k5      逐预算的逐种子明细
  ablation_mnist_curves_k15.csv / k10 / k5      逐预算的逐轮曲线
  ablation_mnist_summary.csv                    各预算的汇总（含最佳轮次）
  ablation_mnist_best_rounds.csv                每组 预算/变体/种子 的最佳准确率与最佳轮次
  ablation_mnist_fig_k15.png / k10 / k5          图：左=每个 seed 的逐轮实际准确率，右=每个 seed 的最佳准确率及其轮次

预计运行时间：4 变体 x 3 种子 x 3 预算 x 40 轮，总计 1440 轮训练。
"""

import os
import csv
import random
import collections
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
from scipy import stats
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input, Conv2D, MaxPooling2D, Flatten
from tensorflow.keras.optimizers import SGD

# ============================================================
# 1. 全局参数（沿用原 notebook）
# ============================================================
NUM_CLIENTS = 20
NUM_CLASSES = 10
IMAGE_SHAPE = (28, 28, 1)

WINDOW_SIZE = 5
WEIGHT_HISTORY = 0.6
WEIGHT_DATA = 0.3
WEIGHT_SYS = 0.1
NUM_SELECT_CLIENTS = 15
LAMBDA_SMOOTH = 0.9
LOCAL_EPOCHS = 1
LEARNING_RATE = 0.01
BATCH_SIZE = 40
NUM_GLOBAL_EPOCHS = 40

# 本次固定单一随机流：只跑 seed=42，单次运行、不做重复实验
# （结论按"单次受控运行"表述，不主张跨种子统计显著性）
SEEDS = [42]

# 作图用哪个种子：只有一个种子，每个预算一张图、4 条线（4 个变体）
PLOT_SEEDS = [42]

# 参与预算：只跑 k=15（每轮从 20 个客户端中选 15 个）
SELECT_BUDGETS = [15]

# 逐变体早停：每个变体以自己的 best-so-far 为基准，连续 PATIENCE_ROUNDS 轮
# 没有刷新自身最佳就“只停这一个变体”，另外三个继续训练（互不影响）。
# 设为 0 则关闭早停，四个变体都跑满 NUM_GLOBAL_EPOCHS 轮（可用于口径对齐）。
PATIENCE_ROUNDS = 3
IMPROVE_EPS = 1e-4

# 严格公平（强制开启，不再提供开关）：四个变体共享
#   1) 完全相同的初始权重（同一个 seed 下只构建一次模型）
#   2) 完全相同的 sys_score（同一 seed 下按 client_id 确定，U(0.7, 1.0)）
# 这样变体之间的差异只来自 DS / AA 的选择与聚合策略本身。
# 注意：客户端选择是算法的一部分，不做"四变体同一轮抽同一批客户端"的强制对齐——
# FedDAAW 按 score 加权抽样、FedAvg 等概率随机抽样，各自的抽样序列本来就该不同。
# 公平性只保证"同一起点（权重 + sys_score + 数据划分）"，不保证"同一条抽样路径"。
# 客户端划分也是同一份 client_data，因此三档之间的差异只来自 NUM_SELECT_CLIENTS。
SHARE_INIT_AND_SYS = True


def set_seeds(seed):
    tf.random.set_seed(seed)
    np.random.seed(seed)
    random.seed(seed)


# ============================================================
# 2. 加载数据
# ============================================================
print("加载MNIST数据...")
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train_original = x_train
y_train_original = y_train

x_test_cnn = x_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0
y_test_categorical = tf.keras.utils.to_categorical(y_test, 10)

print(f"训练集原始形状: {x_train_original.shape}")
print(f"测试集形状: {x_test_cnn.shape}")


# ============================================================
# 3. 工具函数（沿用原 notebook）
# ============================================================
def normalize_scores(scores):
    if len(scores) == 0:
        return np.array([])
    min_score = np.min(scores)
    max_score = np.max(scores)
    if max_score - min_score < 1e-6:
        return np.ones_like(scores) / len(scores)
    return (scores - min_score) / (max_score - min_score)


def build_model():
    model = Sequential([
        Input(shape=IMAGE_SHAPE),
        Conv2D(32, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),
        Conv2D(64, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),
        Flatten(),
        Dropout(0.5),
        Dense(64, activation='relu'),
        Dense(NUM_CLASSES, activation='softmax')
    ])
    model.compile(
        optimizer=SGD(learning_rate=LEARNING_RATE),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


# ============================================================
# 4. Non-IID 划分（沿用原 notebook）
# ============================================================
def load_and_partition_mnist_non_iid(
    num_clients=10,
    client_ratios=None,
    alpha=0.5,
    val_ratio=0.3
):
    x_data = x_train_original
    y_data = y_train_original
    total_samples = len(x_data)

    if client_ratios is None:
        client_ratios = [1.0 / num_clients] * num_clients
    else:
        assert len(client_ratios) == num_clients, "client_ratios 长度必须等于 num_clients"
        assert abs(sum(client_ratios) - 1.0) < 1e-6, "client_ratios 总和必须为 1.0"

    client_sizes = [int(ratio * total_samples) for ratio in client_ratios]
    diff = total_samples - sum(client_sizes)
    for i in range(diff):
        client_sizes[i] += 1

    class_indices = [np.where(y_data == i)[0] for i in range(NUM_CLASSES)]
    used_mask = np.zeros(total_samples, dtype=bool)
    client_data = []

    for i in range(num_clients):
        needed = client_sizes[i]
        class_probs = np.random.dirichlet([alpha] * NUM_CLASSES)
        selected_indices = []
        remaining = needed
        class_order = np.random.permutation(NUM_CLASSES)
        for class_id in class_order:
            n_from_class = int(class_probs[class_id] * needed)
            if n_from_class == 0 and remaining > 0:
                n_from_class = 1
            if n_from_class > remaining:
                n_from_class = remaining
            available_indices = class_indices[class_id][~used_mask[class_indices[class_id]]]
            n_available = len(available_indices)
            if n_available <= n_from_class:
                selected = available_indices
            else:
                selected = np.random.choice(available_indices, size=n_from_class, replace=False)
            selected_indices.extend(selected)
            used_mask[selected] = True
            remaining -= len(selected)
            if remaining <= 0:
                break
        if remaining > 0:
            all_remaining = np.where(~used_mask)[0]
            if len(all_remaining) < remaining:
                selected = all_remaining
            else:
                selected = np.random.choice(all_remaining, size=remaining, replace=False)
            selected_indices.extend(selected)
            used_mask[selected] = True

        assert len(selected_indices) == needed
        selected_indices = np.array(selected_indices)
        np.random.shuffle(selected_indices)

        X_local = x_data[selected_indices]
        y_local = y_data[selected_indices]
        X_local = X_local.reshape(-1, 28, 28, 1).astype('float32') / 255.0

        split = int((1 - val_ratio) * len(X_local))
        client_data.append({
            'X_train': X_local[:split], 'y_train': y_local[:split],
            'X_val': X_local[split:], 'y_val': y_local[split:],
            'class_probs': class_probs
        })

    actual_sizes = [len(d['X_train']) + len(d['X_val']) for d in client_data]
    print(f"[Non-IID 按比例划分完成] 总样本: {total_samples}")
    print(f"客户端数据量 → min: {min(actual_sizes)}, "
          f"max: {max(actual_sizes)}, mean: {np.mean(actual_sizes):.1f}")
    return client_data


# ============================================================
# 5. Client 类（原 notebook 原样，一字未改）
# ============================================================
class Client:
    def __init__(self, client_id, data_dict):
        self.client_id = client_id
        self.X_train = data_dict['X_train']
        self.y_train = data_dict['y_train']
        self.X_val = data_dict['X_val']
        self.y_val = data_dict['y_val']
        self.model = build_model()
        self.acc_history = []
        self.acc_trend = 0.0
        self.data_similarity = 0.0
        self.sys_score = np.random.uniform(0.7, 1.0)

    def local_train(self, global_weights):
        self.model.set_weights(global_weights)
        self.model.fit(
            self.X_train, self.y_train,
            epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE, verbose=0,
            validation_data=(self.X_val, self.y_val)
        )
        return self.model.get_weights()

    def local_evaluate(self):
        y_pred = np.argmax(self.model.predict(self.X_val, verbose=0), axis=1)
        acc = accuracy_score(self.y_val, y_pred)
        self.acc_history.append(acc)
        hist = self.acc_history
        if len(hist) >= WINDOW_SIZE:
            x = np.arange(WINDOW_SIZE)
            slope, _, _, _, _ = stats.linregress(x, hist[-WINDOW_SIZE:])
            self.acc_trend = slope
        elif len(hist) >= 2:
            x = np.arange(len(hist))
            slope, _, _, _, _ = stats.linregress(x, hist)
            self.acc_trend = slope
        else:
            self.acc_trend = 0.0
        return acc

    def get_historical_performance_factor(self):
        return max(0, self.acc_trend) + 0.1

    def get_data_quality_factor(self):
        return max(0.1, min(1.0, self.data_similarity))

    def get_system_factor(self):
        return self.sys_score


# ============================================================
# 6. ServerDA 类（原 notebook 原样，一字未改）
# ============================================================
class ServerDA:
    def __init__(self):
        self.global_model = build_model()
        self.global_weights = self.global_model.get_weights()
        self.global_acc_history = []

    def calculate_selection_score(self, clients):
        scores = []
        for c in clients:
            hist_factor = c.get_historical_performance_factor()
            data_factor = c.get_data_quality_factor()
            sys_factor = c.get_system_factor()
            score = (WEIGHT_HISTORY * hist_factor +
                     WEIGHT_DATA * data_factor +
                     WEIGHT_SYS * sys_factor)
            scores.append(score)
        return np.array(scores)

    def select_clients(self, clients):
        scores = self.calculate_selection_score(clients)
        probs = normalize_scores(scores)
        probs = probs / probs.sum()
        if np.any(np.isnan(probs)) or np.all(probs == 0):
            probs = np.ones(len(clients)) / len(clients)
        indices = np.random.choice(
            len(clients),
            size=min(NUM_SELECT_CLIENTS, len(clients)),
            replace=False,
            p=probs
        )
        selected = [clients[i] for i in indices]
        print(f"DA选择客户端: {[c.client_id for c in selected]}")
        return selected

    def calculate_adaptive_weights(self, selected_clients):
        weights = []
        for c in selected_clients:
            if len(c.acc_history) == 0:
                weights.append(0.1)
            else:
                current = c.acc_history[-1]
                avg = np.mean(c.acc_history)
                smoothed = LAMBDA_SMOOTH * current + (1 - LAMBDA_SMOOTH) * avg
                weights.append(smoothed)
        weights = np.array(weights)
        if np.sum(weights) > 0:
            weights = weights / np.sum(weights)
        else:
            weights = np.ones(len(weights)) / len(weights)
        return weights

    def update_data_similarity(self, selected_clients, client_weights_list):
        if len(selected_clients) == 0:
            return
        global_first_layer = self.global_weights[0].flatten()
        global_norm = np.linalg.norm(global_first_layer)
        for c, cw in zip(selected_clients, client_weights_list):
            client_first_layer = cw[0].flatten()
            client_norm = np.linalg.norm(client_first_layer)
            if global_norm > 1e-8 and client_norm > 1e-8:
                similarity = np.dot(global_first_layer, client_first_layer) / \
                             (global_norm * client_norm)
                c.data_similarity = max(0, min(1, similarity))
            else:
                c.data_similarity = 0.5

    def aggregate(self, selected_clients, client_weights_list):
        if len(selected_clients) == 0:
            return
        adaptive_w = self.calculate_adaptive_weights(selected_clients)
        print(f"DA聚合权重: {adaptive_w}")
        new_weights = []
        for layer_idx in range(len(self.global_weights)):
            weighted_sum = np.zeros_like(self.global_weights[layer_idx])
            for alpha, cw in zip(adaptive_w, client_weights_list):
                weighted_sum += alpha * cw[layer_idx]
            new_weights.append(weighted_sum)
        self.global_weights = new_weights
        self.update_data_similarity(selected_clients, client_weights_list)

    def evaluate(self, test_data=None, test_labels=None):
        if test_data is None:
            test_data = x_test_cnn
            test_labels = y_test_categorical
        self.global_model.set_weights(self.global_weights)
        y_pred = np.argmax(self.global_model.predict(test_data, verbose=0), axis=1)
        if test_labels.ndim > 1:
            y_true = np.argmax(test_labels, axis=1)
        else:
            y_true = test_labels
        acc = accuracy_score(y_true, y_pred)
        self.global_acc_history.append(acc)
        return acc


# ============================================================
# 7. 消融变体 1：去掉动态客户端选择（改随机选）
# ============================================================
class ServerDA_NoDynamicSelect(ServerDA):
    def select_clients(self, clients):
        indices = np.random.choice(
            len(clients),
            size=min(NUM_SELECT_CLIENTS, len(clients)),
            replace=False
        )
        selected = [clients[i] for i in indices]
        print(f"[w/o DS] 选择客户端: {[c.client_id for c in selected]}")
        return selected


# ============================================================
# 8. 消融变体 2：去掉自适应聚合（改简单平均）
# ============================================================
class ServerDA_NoAdaptiveAgg(ServerDA):
    def aggregate(self, selected_clients, client_weights_list):
        if len(selected_clients) == 0:
            return
        new_weights = []
        for layer_idx in range(len(self.global_weights)):
            avg = np.zeros_like(self.global_weights[layer_idx])
            for cw in client_weights_list:
                avg += cw[layer_idx]
            avg = avg / len(client_weights_list)
            new_weights.append(avg)
        self.global_weights = new_weights
        # 保留数据相似度更新，确保消融的只是聚合方式
        self.update_data_similarity(selected_clients, client_weights_list)


# ============================================================
# 9. 消融变体 3：全拆 = FedAvg baseline
# ============================================================
class ServerDA_FedAvgBaseline(ServerDA):
    def select_clients(self, clients):
        indices = np.random.choice(
            len(clients),
            size=min(NUM_SELECT_CLIENTS, len(clients)),
            replace=False
        )
        selected = [clients[i] for i in indices]
        print(f"[FedAvg] 选择客户端: {[c.client_id for c in selected]}")
        return selected

    def aggregate(self, selected_clients, client_weights_list):
        if len(selected_clients) == 0:
            return
        new_weights = []
        for layer_idx in range(len(self.global_weights)):
            avg = np.zeros_like(self.global_weights[layer_idx])
            for cw in client_weights_list:
                avg += cw[layer_idx]
            avg = avg / len(client_weights_list)
            new_weights.append(avg)
        self.global_weights = new_weights


# ============================================================
# 10. 单次运行：一个 seed + 一种参与预算
# ============================================================
def run_ablation(seed, select_budget, client_data):
    """
    client_data 由 main() 按 seed 生成一次，三种参与预算共用同一份划分，
    因此同一 seed 下不同预算之间的差异只来自 NUM_SELECT_CLIENTS。
    """
    global NUM_SELECT_CLIENTS
    NUM_SELECT_CLIENTS = select_budget

    set_seeds(seed)
    print("\n" + "#" * 70)
    print(f"# 消融 | SEED {seed} | 每轮选择 {select_budget}/{NUM_CLIENTS} 客户端 "
          f"| {NUM_GLOBAL_EPOCHS} 轮")
    print("#" * 70)

    variants = collections.OrderedDict([
        ('FedDAAW (Full)', ServerDA),
        ('w/o DS',         ServerDA_NoDynamicSelect),
        ('w/o AA',         ServerDA_NoAdaptiveAgg),
        ('FedAvg',         ServerDA_FedAvgBaseline),
    ])
    cell_tag = {
        'FedDAAW (Full)': 'DS+AA',
        'w/o DS':         'AA_only',
        'w/o AA':         'DS_only',
        'FedAvg':         'none',
    }

    # 每个变体独立的 clients 和 server，互不干扰
    clients_map = {
        name: [Client(i, client_data[i]) for i in range(NUM_CLIENTS)]
        for name in variants
    }
    servers = {name: cls() for name, cls in variants.items()}

    # ---- 严格公平：无条件共享初始权重与 sys_score（不再依赖开关）----
    # 1) 初始权重：先把随机数状态重置到该 seed，再构建一次模型，
    #    四个变体都拷这份权重，避免"谁先 build_model 谁就先吃掉随机数"造成初始化不同。
    set_seeds(seed)
    ref_weights = [w.copy() for w in build_model().get_weights()]

    # 2) sys_score：用独立 RandomState(seed) 生成，按 client_id 一一对应，
    #    与变体创建顺序无关（Client.__init__ 里那个随机值会被这里整体覆盖）。
    shared_sys = np.random.RandomState(seed).uniform(0.7, 1.0, NUM_CLIENTS)

    for name in variants:
        servers[name].global_weights = [w.copy() for w in ref_weights]
        for i, c in enumerate(clients_map[name]):
            c.sys_score = float(shared_sys[i])

    # 自检：确认四个变体的初始权重与 sys_score 完全一致
    _ref0 = servers['FedDAAW (Full)'].global_weights
    _same_w = all(np.array_equal(np.asarray(_ref0[k]), np.asarray(servers[n].global_weights[k]))
                  for n in variants for k in range(len(_ref0)))
    _same_s = all(np.allclose([c.sys_score for c in clients_map[n]],
                              shared_sys) for n in variants)
    print(f"[公平性自检] seed={seed}, k={select_budget}: "
          f"初始权重全一致={_same_w}, sys_score 全一致={_same_s}, "
          f"sys_score 范围=[{shared_sys.min():.3f}, {shared_sys.max():.3f}]")
    assert _same_w and _same_s, '严格公平初始化失败'

    curves = {name: [] for name in variants}
    # ---- 早停状态：每个变体各自独立判断，互不影响 ----
    best_seen = {name: -np.inf for name in variants}      # 该变体自己的历史最佳
    last_improve = {name: 1 for name in variants}         # 该变体最近一次刷新自身最佳的轮次
    stop_round = {name: NUM_GLOBAL_EPOCHS for name in variants}
    stopped = {name: False for name in variants}

    for epoch in range(NUM_GLOBAL_EPOCHS):
        r = epoch + 1
        active = [n for n in variants if not stopped[n]]
        if not active:
            print(f"\n四个变体均已各自早停，第 {r} 轮结束（seed={seed}, k={select_budget}）")
            break

        print(f"\n{'=' * 50}")
        print(f"[seed={seed} | k={select_budget}] 轮次 {r}/{NUM_GLOBAL_EPOCHS}"
              f"（在跑: {', '.join(active)}）")
        print(f"{'=' * 50}")

        for name in active:
            server = servers[name]
            clients = clients_map[name]
            print(f"\n[{name}]")
            selected = server.select_clients(clients)
            weights_list = [c.local_train(server.global_weights) for c in selected]
            for c in selected:
                c.local_evaluate()
            server.aggregate(selected, weights_list)
            acc = server.evaluate()
            curves[name].append(acc)

            # 只用该变体自己的历史最佳来更新计数
            if acc > best_seen[name] + IMPROVE_EPS:
                best_seen[name] = acc
                last_improve[name] = r

            # 逐变体早停：连续 PATIENCE_ROUNDS 轮没有刷新自身 best 就停这一个
            if PATIENCE_ROUNDS > 0 and (r - last_improve[name]) >= PATIENCE_ROUNDS:
                stopped[name] = True
                stop_round[name] = r
                print(f"  --> [{name}] 连续 {PATIENCE_ROUNDS} 轮未刷新自身 best "
                      f"(best={best_seen[name]:.4f} @ r{last_improve[name]})，"
                      f"从第 {r} 轮起停止训练")

        print(f"\n>>> 轮次 {r} 汇总 (seed={seed}, k={select_budget}):")
        for name in variants:
            if curves[name]:
                tag = ' [已早停]' if stopped[name] and r >= stop_round[name] else ''
                print(f"    {name}: {curves[name][-1]:.4f}{tag}")

    # 汇总
    full_best = float(max(curves['FedDAAW (Full)']))
    rows = []
    for name in variants:
        c = curves[name]
        best = float(max(c))
        best_r = int(np.argmax(c)) + 1
        threshold = best * 0.95
        r95 = next((i + 1 for i, a in enumerate(c) if a >= threshold), len(c))
        last5 = float(np.mean(c[-5:])) if len(c) >= 5 else float(c[-1])
        rows.append({
            'select_budget': select_budget,
            'variant': name,
            'cell': cell_tag[name],
            'seed': seed,
            'final_acc': round(float(c[-1]), 6),
            'best_acc': round(best, 6),
            'best_round': best_r,
            'stop_round': stop_round[name],
            'stopped_early': int(stopped[name]),
            'rounds_to_95pct': r95,
            'delta_best_acc_vs_full': round(best - full_best, 6),
            'mean_last5_acc': round(last5, 6),
            'gap_final_best': round((best - float(c[-1])) * 100, 4),
        })
    return rows, curves


# ============================================================
# 11. 可视化：每种参与预算单独一张图（逐种子实际值 + 最佳轮次）
# ============================================================
def plot_ablation(all_curves, select_budget, plot_seed):
    """
    每调用一次 = 该参与预算 + 该种子的一张图：
      线条 = 4 个变体在该种子下的逐轮实际测试准确率（4 条线，不做 mean±std）
      星号 = 该 run 的最佳点
      图下脚注 = 4 个变体在该种子下的最佳轮次与最佳准确率
    """
    style = {
        'FedDAAW (Full)': ('blue',   'o'),
        'w/o DS':         ('orange', 's'),
        'w/o AA':         ('green',  '^'),
        'FedAvg':         ('red',    'D'),
    }

    traces = collections.defaultdict(list)
    for v, sd, r, acc in all_curves:
        traces[(v, sd)].append((r, acc))

    fig, ax = plt.subplots(figsize=(10.8, 6.2))
    notes = []
    for name, (color, marker) in style.items():
        pts = sorted(traces.get((name, plot_seed), []))
        if not pts:
            continue
        x = [p[0] for p in pts]
        y = [p[1] for p in pts]
        ax.plot(x, y, color=color, linestyle='-', linewidth=2.0,
                marker=marker, markersize=4.5, markevery=max(1, len(x) // 10),
                label=name, zorder=4)
        br = x[int(np.argmax(y))]
        ba = max(y)
        ax.plot(br, ba, marker='*', markersize=14, color=color,
                markeredgecolor='black', markeredgewidth=0.8, zorder=6)
        ax.annotate(f'{ba * 100:.2f}% @ r{br}', xy=(br, ba),
                    xytext=(0, 9), textcoords='offset points',
                    fontsize=9, color=color, ha='center', zorder=7,
                    bbox=dict(boxstyle='round,pad=0.18', fc='white',
                              ec=color, alpha=0.85))
        notes.append(f"{name}: best {ba * 100:.2f}% @ r{br}  (final {y[-1] * 100:.2f}%)")

    handles = [plt.Line2D([], [], color=style[n][0], marker=style[n][1],
                          linewidth=2.0, markersize=7, label=n) for n, _ in style.items()]
    handles.append(plt.Line2D([], [], color='black', marker='*', linestyle='none',
                              markersize=13, label='best point'))
    ax.legend(handles=handles, loc='lower right', fontsize=10,
              title=f'variant (seed {plot_seed})', title_fontsize=10, framealpha=0.93)

    ax.set_xlabel('Training Rounds on Non-IID MNIST', fontsize=12)
    ax.set_ylabel('Test Accuracy (actual per-round value, no averaging)', fontsize=11)
    ax.set_title(f'FedDAAW Ablation on Non-IID MNIST  |  seed {plot_seed}  '
                 f'|  k = {select_budget}/{NUM_CLIENTS} clients per round',
                 fontsize=12.5)
    ax.grid(True, linestyle='--', alpha=0.45)
    ax.set_xlim([0, NUM_GLOBAL_EPOCHS + 1])
    all_acc = [a for pts in traces.values() for _, a in pts]
    if all_acc:
        ax.set_ylim([max(0.0, min(all_acc) - 0.03), 1.0])

    fig.subplots_adjust(bottom=0.22, right=0.99)
    footer = (f'Seed {plot_seed}, {NUM_GLOBAL_EPOCHS} scheduled rounds (per-variant early stop allowed).\n'
              'Best round / best accuracy (this seed only):\n'
              + '   '.join(notes))
    fig.text(0.012, 0.015, footer, fontsize=8.8, ha='left', va='bottom',
             linespacing=1.6)

    path = f"results/ablation_mnist_fig_k{select_budget}_seed{plot_seed}.png"
    fig.savefig(path, dpi=160, bbox_inches='tight')
    plt.show()
    plt.close(fig)
    print(f"已写出: {path}")


# ============================================================
# 12. 主程序（三种参与预算一次跑完）
# ============================================================
def main():
    os.makedirs("results", exist_ok=True)

    all_rows = []      # 逐 seed 明细（各预算）
    all_curves = []    # (budget, variant, seed, round, acc)

    fixed_ratios = [0.1, 0.1, 0.1, 0.1, 0.05, 0.05, 0.05, 0.1, 0.1, 0.1,
                    0.005, 0.005, 0.02, 0.03, 0.002, 0.003, 0.002, 0.003, 0.04, 0.04]

    for seed in SEEDS:
        set_seeds(seed)
        print("\n" + "=" * 70)
        print(f"生成 Non-IID 划分 | SEED {seed} | {NUM_CLIENTS} 个客户端")
        print("=" * 70)
        client_data = load_and_partition_mnist_non_iid(
            num_clients=NUM_CLIENTS, alpha=0.5,
            client_ratios=fixed_ratios, val_ratio=0.2
        )

        for select_budget in SELECT_BUDGETS:
            rows, curves = run_ablation(seed, select_budget, client_data)
            all_rows.extend(rows)
            for name, c in curves.items():
                for r, acc in enumerate(c, start=1):
                    all_curves.append((select_budget, name, seed, r, float(acc)))

    # ---- 1) 逐预算的逐 seed 明细 ----
    for select_budget in SELECT_BUDGETS:
        rows_b = [r for r in all_rows if r['select_budget'] == select_budget]
        path = f"results/ablation_mnist_result_k{select_budget}.csv"
        with open(path, "w", newline="") as f:
            w = csv.DictWriter(f, fieldnames=list(rows_b[0].keys()))
            w.writeheader()
            w.writerows(rows_b)
        print(f"\n已写出: {path}")

    # ---- 2) 逐预算的逐轮曲线 ----
    for select_budget in SELECT_BUDGETS:
        path = f"results/ablation_mnist_curves_k{select_budget}.csv"
        with open(path, "w", newline="") as f:
            w = csv.writer(f)
            w.writerow(['select_budget', 'variant', 'seed', 'round', 'test_acc'])
            for b, name, seed, r, acc in all_curves:
                if b == select_budget:
                    w.writerow([b, name, seed, r, acc])
        print(f"已写出: {path}")

    # ---- 3) 汇总（各预算 + 均值 ± 标准差，ddof=0 与论文表口径一致）----
    order = ['FedDAAW (Full)', 'w/o DS', 'w/o AA', 'FedAvg']
    summary_rows = []
    print("\n" + "=" * 78)
    if len(SEEDS) > 1:
        print(f"消融实验汇总（{len(SEEDS)} seeds，{NUM_GLOBAL_EPOCHS} 轮，均值 ± 标准差）")
    else:
        print(f"消融实验汇总（single run, seed={SEEDS[0]}，{NUM_GLOBAL_EPOCHS} 轮；"
              f"无标准差，仅报本次运行值）")
    print("=" * 78)
    for select_budget in SELECT_BUDGETS:
        print(f"\n--- 每轮选择 {select_budget}/{NUM_CLIENTS} 客户端 ---")
        if len(SEEDS) > 1:
            print(f"{'variant':16s} {'final':>14s} {'best':>14s} "
                  f"{'r95':>12s} {'last5':>8s} {'gap':>7s}")
        else:
            print(f"{'variant':16s} {'final':>14s} {'best':>14s} "
                  f"{'r95':>12s} {'last5':>8s} {'gap':>7s}")
        for name in order:
            v = [r for r in all_rows
                 if r['select_budget'] == select_budget and r['variant'] == name]
            fin = np.array([x['final_acc'] for x in v])
            bst = np.array([x['best_acc'] for x in v])
            r95 = np.array([x['rounds_to_95pct'] for x in v])
            l5 = np.array([x['mean_last5_acc'] for x in v])
            gap = np.array([x['gap_final_best'] for x in v])
            if len(v) > 1:
                print(f"{name:16s} {fin.mean()*100:7.2f}±{fin.std()*100:4.2f}  "
                      f"{bst.mean()*100:7.2f}±{bst.std()*100:4.2f}  "
                      f"{r95.mean():5.2f}±{r95.std():4.2f}  "
                      f"{l5.mean()*100:6.2f}  {gap.mean():6.2f}")
            else:
                # 单次运行：标准差无意义，只打印该次运行的值
                print(f"{name:16s} {fin.mean()*100:7.2f}(n=1)   "
                      f"{bst.mean()*100:7.2f}(n=1)   "
                      f"{r95.mean():5.2f}        "
                      f"{l5.mean()*100:6.2f}  {gap.mean():6.2f}")
            summary_rows.append({
                'select_budget': select_budget,
                'variant': name,
                'cell': v[0]['cell'],
                'n_seeds': len(v),
                'final_mean': round(fin.mean() * 100, 4),
                'final_std': round(fin.std() * 100, 4),
                'best_mean': round(bst.mean() * 100, 4),
                'best_std': round(bst.std() * 100, 4),
                'best_round_mean': round(float(np.mean([x['best_round'] for x in v])), 2),
                'best_round_per_seed': '/'.join(str(int(x['best_round'])) for x in v),
                'r95_mean': round(r95.mean(), 4),
                'r95_std': round(r95.std(), 4),
                'last5_mean': round(l5.mean() * 100, 4),
                'gap_final_best_mean': round(gap.mean(), 4),
            })

    with open("results/ablation_mnist_summary.csv", "w", newline="") as f:
        w = csv.DictWriter(f, fieldnames=list(summary_rows[0].keys()))
        w.writeheader()
        w.writerows(summary_rows)
    print("\n已写出: results/ablation_mnist_summary.csv")

    # ---- 4) 最佳轮次表：每组 预算/变体/种子 的 best_acc 与 best_round ----
    best_rows = []
    for r in all_rows:
        best_rows.append({
            'select_budget': r['select_budget'],
            'variant': r['variant'],
            'seed': r['seed'],
            'best_acc': r['best_acc'],
            'best_round': r['best_round'],
            'final_acc': r['final_acc'],
        })
    with open("results/ablation_mnist_best_rounds.csv", "w", newline="") as f:
        w = csv.DictWriter(f, fieldnames=['select_budget', 'variant', 'seed',
                                          'best_acc', 'best_round', 'final_acc'])
        w.writeheader()
        w.writerows(best_rows)
    print("\n已写出: results/ablation_mnist_best_rounds.csv")

    # 打印最佳轮次汇总（逐预算下每个变体的 best_round 均值与逐种子值）
    print("\n" + "=" * 78)
    print("最佳准确率出现轮次（best_round）")
    print("=" * 78)
    for select_budget in SELECT_BUDGETS:
        print(f"\n--- 每轮选择 {select_budget}/{NUM_CLIENTS} 客户端 ---")
        for name in order:
            v = [r for r in all_rows
                 if r['select_budget'] == select_budget and r['variant'] == name]
            rds = [int(x['best_round']) for x in v]
            print(f"  {name:16s} mean r{np.mean(rds):4.1f} | per-seed "
                  + " / ".join(f"r{x}" for x in rds))

    # ---- 5) 出图：每个 (预算, 种子) 一张，共 len(SELECT_BUDGETS)*len(PLOT_SEEDS) 张 ----
    for select_budget in SELECT_BUDGETS:
        for plot_seed in PLOT_SEEDS:
            sub = [(c[1], c[2], c[3], c[4]) for c in all_curves
                   if c[0] == select_budget and c[2] == plot_seed]
            if not sub:
                print(f"跳过：k={select_budget}, seed={plot_seed} 无曲线数据")
                continue
            plot_ablation(sub, select_budget, plot_seed)


if __name__ == "__main__":
    main()
```
