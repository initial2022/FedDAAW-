```python
"""
Federated Learning Comparison: FedDAAW vs FedGAC vs FedAA vs FedClust
Dataset: California Housing (Regression, Non-IID distribution)

改写说明（在附件2基础上）：
  - 保留附件2的四个对比算法：FedDAAW / FedGAC / FedAA / FedClust
  - 数据集由 MNIST 替换为 California Housing（回归任务，8 特征 -> 1 房价）
  - 网络结构与附件1一致：回归 MLP(128 -> 64 -> 1)，损失 MSE，指标 MAE
  - FedDAAW 按附件1思路：MSE 历史趋势、数据相似度(首层权重余弦)、系统分；
    自适应聚合权重按逆 MSE 计算
  - 其余算法仅把分类信号（准确率 / 类别聚类）替换为回归口径（MSE / 参数特征）

—— 本版（paper-aligned）对三个基线的实现做了与各自论文对齐的修改 ——
  FedGAC：CLP 阶段按数据质量选择（原为随机）；AILM 作用于个性化模型；
          新增个性化聚合（FEDGAC_PERSONAL_BETA）、按数据质量调整本地 epoch
          （FEDGAC_QUALITY_BASE / FEDGAC_QUALITY_EPOCH_GAIN）、参数压缩 top-k
          （FEDGAC_COMPRESS_RATIO）并据此核算上行开销。
  FedAA：选择由随机改为按模型参数距离取 top M%（FEDAA_M_PERCENT）；
          奖励改为服务器留出验证集（取自训练集，FEDAA_SERVER_VAL_RATIO），
          不再使用测试集，消除原实现中奖励与评估集重叠的问题。
  FedClust：聚类特征由倒数第二个 Dense 的 kernel 改为论文规定的末层权重与偏置；
          新增新客户端并入接口（Algorithm 2，FEDCLUST_ENABLE_NEWCOMER）。
  说明：论文中无法在本实验条件下完整实现的机制，均以一个显式常量参数代替，
        参数与论文条目的对应关系写在各自类与方法的注释中。
"""

import os
import csv
import random
import collections
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler
from scipy import stats
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (Dense, Dropout, Input, Concatenate)
from tensorflow.keras.optimizers import SGD, Adam
from tensorflow.keras import layers
import warnings
warnings.filterwarnings('ignore')
# ============================================================
# 1. 全局参数
# ============================================================
NUM_CLIENTS = 20
INPUT_DIM = 8          # California Housing 共 8 个特征
OUTPUT_DIM = 1         # 回归任务：输出房价 1 维

# FedDAAW 客户端选择 / 自适应聚合相关
WINDOW_SIZE = 3
WEIGHT_HISTORY = 0.6
WEIGHT_DATA = 0.3
WEIGHT_SYS = 0.1
NUM_SELECT_CLIENTS = 15
LAMBDA_SMOOTH = 0.9

LOCAL_EPOCHS = 1
LEARNING_RATE = 0.01
BATCH_SIZE = 40
NUM_GLOBAL_EPOCHS = 25

SEEDS = [42, 123, 2024]

# ========== FedGAC ==========
FEDGAC_CLP_ROUNDS = 3
FEDGAC_AILM_ALPHA = 0.1

# ========== FedClust ==========
FEDCLUST_TAU_QUANTILE = 0.5

# ========== FedAA (DDPG) ==========
FEDAA_HIDDEN = 64
FEDAA_LR_ACTOR = 1e-4
FEDAA_LR_CRITIC = 1e-3
FEDAA_GAMMA = 0.9
FEDAA_TAU = 0.01
FEDAA_BUFFER = 500
FEDAA_BATCH = 8
FEDAA_NOISE = 0.1
FEDAA_UPD_ITERS = 3

# ========== 论文对齐用的替代参数（论文机制在本实验条件下的显式取值）==========
FEDGAC_QUALITY_BASE = 0.5        # 数据质量调参：本地 epoch 乘子的基数项
FEDGAC_QUALITY_EPOCH_GAIN = 1.0  # 数据质量调参：质量因子对 epoch 的增益
FEDGAC_PERSONAL_BETA = 0.5       # 个性化聚合：个性化模型向全局模型靠拢的比例
FEDGAC_COMPRESS_RATIO = 0.1      # 参数压缩：上行 top-k 保留比例
FEDGAC_CLP_STRATEGY = 'quality'  # CLP 阶段的选择准则（quality 或 random）
FEDAA_M_PERCENT = 0.75           # FedAA：按参数邻近度选取的 top M% 客户端比例
FEDAA_SERVER_VAL_RATIO = 0.05    # FedAA：服务器留出验证集占训练集比例（奖励来源）
FEDCLUST_ENABLE_NEWCOMER = False # FedClust：新客户端并入开关（静态实验中无新客户端）

# 早停
PATIENCE_ROUNDS = 3
IMPROVE_EPS = 1e-4

# 固定客户端数据比例（Non-IID 分布）
FIXED_RATIOS = [0.1, 0.1, 0.1, 0.1, 0.05, 0.05, 0.05, 0.1, 0.1, 0.1,
                0.005, 0.005, 0.02, 0.03, 0.002, 0.003, 0.002, 0.003, 0.04, 0.04]
def set_seeds(seed):
    tf.random.set_seed(seed)
    np.random.seed(seed)
    random.seed(seed)


# ============================================================
# 2. 加载 California Housing 数据（回归）
# ============================================================
def load_california_housing_data():
    """加载并预处理 California Housing：特征/标签标准化，80/20 划分训练/测试"""
    try:
        housing = fetch_california_housing()
        X, y = housing.data, housing.target
    except Exception:
        # 离线回退：本地 csv，最后一列为房价
        data = np.loadtxt('cal_housing.data', delimiter=',', skiprows=1)
        X, y = data[:, :-1], data[:, -1]

    scaler_X = StandardScaler()
    X_scaled = scaler_X.fit_transform(X)

    scaler_y = StandardScaler()
    y_scaled = scaler_y.fit_transform(y.reshape(-1, 1)).flatten()

    split_idx = int(0.8 * len(X_scaled))
    X_train, X_test = X_scaled[:split_idx], X_scaled[split_idx:]
    y_train, y_test = y_scaled[:split_idx], y_scaled[split_idx:]

    # 【与 FedAA 论文对齐】服务器留出验证集，供 DDPG 奖励使用。
    # 取自训练集，与测试集完全不重叠；原实现从测试集切 1000 条作奖励，
    # 与评估集重叠约 10%，会高估 FedAA 的表现。
    n_server = int(FEDAA_SERVER_VAL_RATIO * len(X_train))
    X_server, y_server = X_train[:n_server], y_train[:n_server]
    X_train, y_train = X_train[n_server:], y_train[n_server:]

    return (X_train, y_train), (X_test, y_test), (X_server, y_server), scaler_y


print("Loading California Housing dataset...")
(X_train, y_train), (X_test, y_test), (X_SERVER, Y_SERVER), y_scaler = load_california_housing_data()
print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"Training labels shape: {y_train.shape}")
print(f"Test labels shape: {y_test.shape}")


# ============================================================
# 3. 基础工具
# ============================================================
def build_model():
    """回归 MLP（与附件1一致）：128 -> 64 -> 1"""
    m = Sequential([
        Input(shape=(INPUT_DIM,)),
        Dense(128, activation='relu'),
        Dropout(0.3),
        Dense(64, activation='relu'),
        Dropout(0.3),
        Dense(OUTPUT_DIM)  # 回归：末层不使用激活函数
    ])
    m.compile(optimizer=Adam(learning_rate=LEARNING_RATE),
              loss='mse',
              metrics=['mae'])
    return m


def normalize_scores(scores):
    """将分数归一化到 [0, 1]"""
    if len(scores) == 0:
        return np.array([])
    mn, mx = np.min(scores), np.max(scores)
    if mx - mn < 1e-6:
        return np.ones_like(scores) / len(scores)
    return (scores - mn) / (mx - mn)


# ============================================================
# 4. Non-IID 划分（按固定比例随机分配；回归任务不按类别划分）
# ============================================================
def load_and_partition_housing_non_iid(num_clients=10, client_ratios=None,
                                      alpha=0.5, val_ratio=0.2):
    total = len(X_train)
    if client_ratios is None:
        client_ratios = np.random.dirichlet([alpha] * num_clients)
    assert len(client_ratios) == num_clients, "client_ratios 长度需等于 num_clients"
    assert abs(sum(client_ratios) - 1.0) < 1e-6, "client_ratios 之和需为 1.0"

    client_sizes = [int(r * total) for r in client_ratios]
    diff = total - sum(client_sizes)
    for i in range(diff):
        client_sizes[i] += 1

    all_indices = np.arange(total)
    np.random.shuffle(all_indices)

    partitions, start = [], 0
    for size in client_sizes:
        partitions.append(all_indices[start:start + size])
        start += size

    client_data = []
    for i in range(num_clients):
        selected = partitions[i]
        np.random.shuffle(selected)
        X_local = X_train[selected]
        y_local = y_train[selected]
        split = int((1 - val_ratio) * len(X_local))
        client_data.append({
            'X_train': X_local[:split], 'y_train': y_local[:split],
            'X_val': X_local[split:], 'y_val': y_local[split:]
        })

    actual = [len(d['X_train']) + len(d['X_val']) for d in client_data]
    print(f"[Proportional random partition completed] Total={total}, "
          f"min={min(actual)}, max={max(actual)}, mean={np.mean(actual):.1f}")
    for i in range(min(10, num_clients)):
        d = client_data[i]
        print(f"Client {i}: Samples={len(d['y_train']) + len(d['y_val'])}, "
              f"Label mean={np.mean(d['y_train']):.3f}±{np.std(d['y_train']):.3f}, "
              f"range=[{np.min(d['y_train']):.3f}, {np.max(d['y_train']):.3f}]")
    return client_data
# ============================================================
# 5. FedDAAW 客户端（按附件1：MSE 历史趋势 / 数据相似度 / 系统分）
# ============================================================
class Client:
    """FedDAAW 客户端：维护历史 MSE 趋势、数据相似度、系统分"""
    def __init__(self, client_id, data_dict):
        self.client_id = client_id
        self.X_train = data_dict['X_train']
        self.y_train = data_dict['y_train']
        self.X_val = data_dict['X_val']
        self.y_val = data_dict['y_val']
        self.model = build_model()
        self.mse_history = []
        self.mse_trend = 0.0
        self.data_similarity = 0.0
        self.sys_score = np.random.uniform(0.7, 1.0)

    def local_train(self, global_weights):
        self.model.set_weights(global_weights)
        self.model.fit(self.X_train, self.y_train,
                       epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE, verbose=0)
        return self.model.get_weights()

    def local_evaluate(self):
        y_pred = self.model.predict(self.X_val, verbose=0).flatten()
        mse = mean_squared_error(self.y_val, y_pred)
        self.mse_history.append(mse)

        hist = self.mse_history
        if len(hist) >= WINDOW_SIZE:
            slope, *_ = stats.linregress(np.arange(WINDOW_SIZE), hist[-WINDOW_SIZE:])
            self.mse_trend = -slope   # MSE 下降 -> 趋势为正
        elif len(hist) >= 2:
            slope, *_ = stats.linregress(np.arange(len(hist)), hist)
            self.mse_trend = -slope
        else:
            self.mse_trend = 0.0
        return mse

    # --- 选择信号接口（FedDAAW 专用） ---
    def get_historical_performance_factor(self):
        return max(0, self.mse_trend) + 0.1

    def get_data_quality_factor(self):
        return max(0.1, min(1.0, self.data_similarity))

    def get_system_factor(self):
        return self.sys_score


# ============================================================
# 6. FedGAC 客户端
# 选择信号：本地验证 MSE（越低越优先）；训练支持 AILM 自定义初始化
# ============================================================
class FedGACClient:
    def __init__(self, client_id, data_dict):
        self.client_id = client_id
        self.X_train = data_dict['X_train']
        self.y_train = data_dict['y_train']
        self.X_val = data_dict['X_val']
        self.y_val = data_dict['y_val']
        self.model = build_model()

    def get_local_validation_mse(self, weights):
        """本地验证 MSE，供 CLP 后按低 MSE 优先选择客户端"""
        self.model.set_weights(weights)
        y_pred = self.model.predict(self.X_val, verbose=0).flatten()
        return float(mean_squared_error(self.y_val, y_pred))

    def local_train_with_init(self, init_weights, epochs=None):
        """epochs 由服务器按数据质量给出（论文：Training parameters are dynamically
        adjusted according to dataset quality），缺省时退回 LOCAL_EPOCHS"""
        self.model.set_weights(init_weights)
        self.model.fit(self.X_train, self.y_train,
                       epochs=LOCAL_EPOCHS if epochs is None else int(epochs),
                       batch_size=BATCH_SIZE, verbose=0)
        return self.model.get_weights()

    def sample_count(self):
        return len(self.X_train) + len(self.X_val)


# ============================================================
# 7. FedAA 客户端
# 选择信号：无（随机选）；暴露参数向量供服务器 DDPG 组状态
# ============================================================
class FedAAClient:
    def __init__(self, client_id, data_dict):
        self.client_id = client_id
        self.X_train = data_dict['X_train']
        self.y_train = data_dict['y_train']
        self.X_val = data_dict['X_val']
        self.y_val = data_dict['y_val']
        self.model = build_model()

    def get_flat_params(self, weights=None):
        if weights is not None:
            self.model.set_weights(weights)
        return np.concatenate([w.flatten() for w in self.model.get_weights()]).astype(np.float64)

    def local_train(self, global_weights):
        self.model.set_weights(global_weights)
        self.model.fit(self.X_train, self.y_train,
                       epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE, verbose=0)
        return self.model.get_weights()


# ============================================================
# 8. FedClust 客户端
# 选择信号：倒数第二层 Dense 的 kernel（聚类特征）+ 簇归属
# ============================================================
class FedClustClient:
    def __init__(self, client_id, data_dict):
        self.client_id = client_id
        self.X_train = data_dict['X_train']
        self.y_train = data_dict['y_train']
        self.X_val = data_dict['X_val']
        self.y_val = data_dict['y_val']
        self.model = build_model()
        self.cluster_id = None

    def get_clustering_feature(self):
        """【与 FedClust 论文 4.1 对齐】只用末层的权重与偏置作为客户端特征。
        论文原文：FedClust utilizes only the final layer weights instead of the full
        model weights ... specifically the weights and bias of the final layer。
        原实现取倒数第二个 Dense 的 kernel，与论文不符。
        """
        kernel, bias = self.model.layers[-1].get_weights()
        return np.concatenate([kernel.flatten(), bias.flatten()]).astype(np.float64)

    def set_cluster_id(self, cid):
        self.cluster_id = cid

    def get_cluster_id(self):
        return self.cluster_id

    def local_train_for_clustering(self, global_weights, epochs=2):
        """首轮聚类前从全局模型出发训练，用于提取聚类特征"""
        self.model.set_weights(global_weights)
        self.model.fit(self.X_train, self.y_train,
                       epochs=epochs, batch_size=BATCH_SIZE, verbose=0)
        return self.model.get_weights()

    def local_train_from_cluster(self, cluster_weights):
        """从所属簇的模型出发训练"""
        self.model.set_weights(cluster_weights)
        self.model.fit(self.X_train, self.y_train,
                       epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE, verbose=0)
        return self.model.get_weights()

    def sample_count(self):
        return len(self.X_train) + len(self.X_val)
# ============================================================
# 9. Server: FedDAAW（按附件1思路）
# ============================================================
class ServerDA:
    def __init__(self):
        self.global_model = build_model()
        self.global_weights = self.global_model.get_weights()
        self.global_mse_history = []
        self.n_params = sum(w.size for w in self.global_weights)
        self.comm_params = 0
        self.comm_bytes = 0

    def calculate_selection_score(self, clients):
        return np.array([
            WEIGHT_HISTORY * c.get_historical_performance_factor() +
            WEIGHT_DATA * c.get_data_quality_factor() +
            WEIGHT_SYS * c.get_system_factor()
            for c in clients
        ])

    def select_clients(self, clients):
        scores = self.calculate_selection_score(clients)
        probs = normalize_scores(scores)
        probs = probs / probs.sum()
        if np.any(np.isnan(probs)) or np.all(probs == 0):
            probs = np.ones(len(clients)) / len(clients)
        idx = np.random.choice(len(clients),
                               size=min(NUM_SELECT_CLIENTS, len(clients)),
                               replace=False, p=probs)
        selected = [clients[i] for i in idx]
        print(f"DA selected clients: {[c.client_id for c in selected]}")
        return selected

    def calculate_adaptive_weights(self, selected):
        """附件1思路：逆 MSE 作为聚合权重（MSE 越低权重越高）"""
        weights = []
        for c in selected:
            cur = c.mse_history[-1] if c.mse_history else 1.0
            cur = max(cur, 1e-8)
            weights.append(1.0 / cur)
        w = np.array(weights)
        w = normalize_scores(w)
        return w / w.sum() if w.sum() > 0 else np.ones(len(w)) / len(w)

    def update_data_similarity(self, selected, client_weights_list):
        if not selected:
            return
        gf = self.global_weights[0].flatten()
        gn = np.linalg.norm(gf)
        for c, cw in zip(selected, client_weights_list):
            cf = cw[0].flatten()
            cn = np.linalg.norm(cf)
            if gn > 1e-8 and cn > 1e-8:
                c.data_similarity = max(0.0, min(1.0, np.dot(gf, cf) / (gn * cn)))
            else:
                c.data_similarity = 0.5

    def aggregate(self, selected, client_weights_list):
        if not selected:
            return
        adaptive_w = self.calculate_adaptive_weights(selected)
        print(f"DA aggregation weights: {adaptive_w}")
        new_w = []
        for li in range(len(self.global_weights)):
            s = np.zeros_like(self.global_weights[li])
            for a, cw in zip(adaptive_w, client_weights_list):
                s += a * cw[li]
            new_w.append(s)
        self.global_weights = new_w
        self.update_data_similarity(selected, client_weights_list)
        self.comm_params += len(selected) * self.n_params
        self.comm_bytes += len(selected) * self.n_params * 4 * 2

    def evaluate(self, test_data=None, test_labels=None):
        if test_data is None:
            test_data, test_labels = X_test, y_test
        self.global_model.set_weights(self.global_weights)
        y_pred = self.global_model.predict(test_data, verbose=0).flatten()
        mse = mean_squared_error(test_labels, y_pred)
        self.global_mse_history.append(mse)
        return mse


# ============================================================
# 10. Server: FedGAC（基于本地验证 MSE 的选择）
# ============================================================
class ServerFedGAC:
    def __init__(self):
        self.global_model = build_model()
        self.global_weights = self.global_model.get_weights()
        self.global_mse_history = []
        self.local_history = {}
        self.round = 0
        self.n_params = sum(w.size for w in self.global_weights)
        self.comm_params = 0
        self.comm_bytes = 0
        # 【与 FedGAC 论文对齐】个性化聚合的客户端个性化模型，以及数据质量因子
        self.personal_weights = {}
        self.quality = {}

    def _data_quality(self, clients):
        """数据质量因子：本地验证 MSE 的倒数归一化到 [0,1]。
        论文只说明按数据质量调整训练参数而未给出量化口径，此处以本地验证 MSE
        的倒数作为质量的代理，并以 FEDGAC_QUALITY_BASE / FEDGAC_QUALITY_EPOCH_GAIN
        把质量映射为本地 epoch 数。"""
        mses = np.array([c.get_local_validation_mse(self.global_weights) for c in clients],
                        dtype=np.float64)
        inv = 1.0 / (mses + 1e-6)
        lo, hi = inv.min(), inv.max()
        if hi - lo < 1e-9:
            return np.ones_like(inv)
        return (inv - lo) / (hi - lo)

    def select_clients(self, clients):
        self.round += 1
        mses = np.array([c.get_local_validation_mse(self.global_weights) for c in clients])
        q = self._data_quality(clients)
        self.quality = {c.client_id: float(q[i]) for i, c in enumerate(clients)}
        if self.round <= FEDGAC_CLP_ROUNDS:
            # 【与论文对齐】论文要求 strategically select clients during CLP，
            # 原实现在 CLP 阶段随机选，与此不符；改为按数据质量降序（FEDGAC_CLP_STRATEGY）
            if FEDGAC_CLP_STRATEGY == 'quality':
                idx = np.argsort(-q)[:min(NUM_SELECT_CLIENTS, len(clients))]
            else:
                idx = np.random.choice(len(clients),
                                       size=min(NUM_SELECT_CLIENTS, len(clients)),
                                       replace=False)
        else:
            # CLP 后：按本地验证 MSE 升序（低 MSE 优先）top-k
            idx = np.argsort(mses)[:min(NUM_SELECT_CLIENTS, len(clients))]
        selected = [clients[i] for i in idx]
        print(f"FedGAC selected clients: {[c.client_id for c in selected]}")
        return selected

    def _compress(self, weights):
        """【与 FedGAC 论文对齐】参数压缩：按绝对值保留 top-k（FEDGAC_COMPRESS_RATIO），
        其余置零。论文只说明 employs parameter compression techniques 而未给出压缩率，
        此处以 10% 的稠密度作为替代参数。"""
        out = []
        for w in weights:
            f = w.flatten()
            k = max(1, int(FEDGAC_COMPRESS_RATIO * f.size))
            thr = np.partition(np.abs(f), -k)[-k]
            sparse = np.where(np.abs(f) >= thr, f, 0.0)
            out.append(sparse.reshape(w.shape))
        return out

    def local_train_with_ailm(self, selected):
        """AILM 初始化 + 个性化起点 + 数据质量调参 + 参数压缩"""
        weights_list = []
        for c in selected:
            cid = c.client_id
            # 个性化聚合：以该客户端个性化模型为起点（无则退回全局模型）
            base = self.personal_weights.get(cid, self.global_weights)
            g = self.global_weights
            if cid in self.local_history:
                l = self.local_history[cid]
                init_w = [(1 - FEDGAC_AILM_ALPHA) * base[i] + FEDGAC_AILM_ALPHA * l[i]
                          for i in range(len(g))]
            else:
                init_w = base
            # 数据质量调参：epoch = LOCAL_EPOCHS * (BASE + GAIN * quality)
            q = self.quality.get(cid, 0.5)
            epochs = max(1, int(round(LOCAL_EPOCHS * (FEDGAC_QUALITY_BASE
                                                     + FEDGAC_QUALITY_EPOCH_GAIN * q))))
            cw = c.local_train_with_init(init_w, epochs=epochs)
            self.local_history[cid] = [w.copy() for w in cw]
            weights_list.append(self._compress(cw))
        return weights_list

    def aggregate(self, selected, client_weights_list):
        if not selected:
            return
        sizes = np.array([c.sample_count() for c in selected], dtype=np.float64)
        sizes = sizes / sizes.sum()
        new_w = []
        for li in range(len(self.global_weights)):
            acc = np.zeros_like(self.global_weights[li])
            for a, cw in zip(sizes, client_weights_list):
                acc += a * cw[li]
            new_w.append(acc)
        self.global_weights = new_w

        # 【与 FedGAC 论文对齐】个性化聚合：客户端个性化模型按 FEDGAC_PERSONAL_BETA
        # 从自身状态向全局模型靠拢；论文只说明 personalized aggregation strategy，
        # 混合比例以常量参数代替。
        for c in selected:
            cid = c.client_id
            p = self.personal_weights.get(cid)
            if p is None:
                self.personal_weights[cid] = [w.copy() for w in self.global_weights]
            else:
                self.personal_weights[cid] = [
                    (1 - FEDGAC_PERSONAL_BETA) * p[i]
                    + FEDGAC_PERSONAL_BETA * self.global_weights[i]
                    for i in range(len(p))
                ]
            c.model.set_weights(self.personal_weights[cid])

        # 压缩后的上行开销：top-k 数值（4 字节/值，索引开销不计入）
        n_up = int(self.n_params * FEDGAC_COMPRESS_RATIO) + 1
        self.comm_params += len(selected) * n_up
        self.comm_bytes += len(selected) * n_up * 4

    def evaluate(self, test_data=None, test_labels=None):
        if test_data is None:
            test_data, test_labels = X_test, y_test
        self.global_model.set_weights(self.global_weights)
        y_pred = self.global_model.predict(test_data, verbose=0).flatten()
        mse = mean_squared_error(test_labels, y_pred)
        self.global_mse_history.append(mse)
        return mse


# ============================================================
# 11. Server: FedAA（随机选 + DDPG 自适应聚合；奖励 = -MSE）
# ============================================================
class ServerFedAA:
    def __init__(self):
        self.global_model = build_model()
        self.global_weights = self.global_model.get_weights()
        self.global_mse_history = []
        self.n_params = sum(w.size for w in self.global_weights)
        self.comm_params = 0
        self.comm_bytes = 0
        self.m = NUM_SELECT_CLIENTS

        self.actor = self._build_actor()
        self.critic = self._build_critic()
        self.actor_target = tf.keras.models.clone_model(self.actor)
        self.actor_target.set_weights(self.actor.get_weights())
        self.critic_target = tf.keras.models.clone_model(self.critic)
        self.critic_target.set_weights(self.critic.get_weights())
        self.actor_opt = Adam(FEDAA_LR_ACTOR)
        self.critic_opt = Adam(FEDAA_LR_CRITIC)
        self.buffer = []

    def _build_actor(self):
        inp = Input(shape=(self.m,))
        x = Dense(FEDAA_HIDDEN, activation='relu')(inp)
        x = Dense(FEDAA_HIDDEN, activation='relu')(x)
        return Model(inp, Dense(self.m)(x))

    def _build_critic(self):
        s_in = Input(shape=(self.m,))
        a_in = Input(shape=(self.m,))
        x = Concatenate()([s_in, a_in])
        x = Dense(FEDAA_HIDDEN, activation='relu')(x)
        x = Dense(FEDAA_HIDDEN, activation='relu')(x)
        return Model([s_in, a_in], Dense(1)(x))

    def select_clients(self, clients):
        """【与 FedAA 论文对齐】按模型参数距离取最接近的 top M% 客户端。
        论文：selects the top M% of clients based on model parameter proximity,
        thus protecting the aggregation against adversarial influences。
        原实现为随机抽样。M 由 FEDAA_M_PERCENT 给出；若 top M% 多于本轮参与名额，
        按距离升序截断到 NUM_SELECT_CLIENTS。"""
        reps = []
        for c in clients:
            c.model.set_weights(self.global_weights)
            reps.append(c.get_flat_params())
        n = len(reps)
        D = np.zeros((n, n))
        for i in range(n):
            for j in range(i + 1, n):
                d = np.linalg.norm(reps[i] - reps[j])
                D[i, j] = D[j, i] = d
        prox = D.sum(axis=1) / max(1, n - 1)   # 与其余客户端的平均参数距离，越小越接近
        k = max(2, int(round(FEDAA_M_PERCENT * n)))
        idx = np.argsort(prox)[:k]
        idx = idx[:min(NUM_SELECT_CLIENTS, len(idx))]
        selected = [clients[i] for i in idx]
        print(f"FedAA selected clients (top {FEDAA_M_PERCENT:.0%} by param proximity): "
              f"{[c.client_id for c in selected]}")
        return selected

    def _state(self, selected, client_weights_list):
        reps = []
        for c, w in zip(selected, client_weights_list):
            c.model.set_weights(w)
            reps.append(c.get_flat_params())
        n = len(reps)
        if n < 2:
            return np.zeros(n, dtype=np.float32)
        M = np.zeros((n, n))
        for i in range(n):
            for j in range(n):
                M[i, j] = np.linalg.norm(reps[i] - reps[j])
        s = M.sum(axis=1)
        s = (s - s.mean()) / (s.std() + 1e-8)
        return s.astype(np.float32)

    def _soft_update(self, target, src):
        tw, sw = target.get_weights(), src.get_weights()
        target.set_weights([FEDAA_TAU * sw[i] + (1 - FEDAA_TAU) * tw[i]
                            for i in range(len(tw))])

    def _ddpg_update(self):
        if len(self.buffer) < FEDAA_BATCH:
            return
        for _ in range(FEDAA_UPD_ITERS):
            batch = random.sample(self.buffer, FEDAA_BATCH)
            s = np.stack([b[0] for b in batch]).astype(np.float32)
            a = np.stack([b[1] for b in batch]).astype(np.float32)
            r = np.array([[b[2]] for b in batch], dtype=np.float32)
            s2 = np.stack([b[3] for b in batch]).astype(np.float32)
            a2 = tf.nn.softmax(self.actor_target(s2), axis=-1).numpy()
            y = r + FEDAA_GAMMA * self.critic_target([s2, a2]).numpy()
            with tf.GradientTape() as tape:
                q = self.critic([s, a], training=True)
                c_loss = tf.reduce_mean(tf.square(y - q))
            g = tape.gradient(c_loss, self.critic.trainable_variables)
            self.critic_opt.apply_gradients(zip(g, self.critic.trainable_variables))
            with tf.GradientTape() as tape:
                a_pred = tf.nn.softmax(self.actor(s, training=True), axis=-1)
                a_loss = -tf.reduce_mean(self.critic([s, a_pred], training=True))
            g = tape.gradient(a_loss, self.actor.trainable_variables)
            self.actor_opt.apply_gradients(zip(g, self.actor.trainable_variables))
        self._soft_update(self.actor_target, self.actor)
        self._soft_update(self.critic_target, self.critic)

    def aggregate(self, selected, client_weights_list):
        if not selected:
            return
        s = self._state(selected, client_weights_list)
        logits = self.actor(s[None, :], training=False).numpy()[0]
        w = np.exp(logits - logits.max())
        w = w / (w.sum() + 1e-12)
        w = w + np.random.randn(len(w)) * FEDAA_NOISE
        w = np.clip(w, 0, None)
        w = w / (w.sum() + 1e-12)

        new_w = []
        for li in range(len(self.global_weights)):
            v = np.zeros_like(self.global_weights[li])
            for a, cw in zip(w, client_weights_list):
                v += a * cw[li]
            new_w.append(v)
        self.global_weights = new_w
        self.comm_params += len(selected) * self.n_params
        self.comm_bytes += len(selected) * self.n_params * 4 * 2

        # 【与 FedAA 论文对齐】奖励由验证集表现给出（论文：a reward mechanism guided by
        # validation set performance），此处用服务器留出验证集的 MSE，不使用测试集。
        reward = -self.evaluate_on(X_SERVER, Y_SERVER)
        self.buffer.append((s, w, reward, s))
        if len(self.buffer) > FEDAA_BUFFER:
            self.buffer.pop(0)
        self._ddpg_update()

    def evaluate_on(self, data, labels):
        """在指定数据集上评估当前全局模型（用于验证集奖励，不写入评估历史）"""
        self.global_model.set_weights(self.global_weights)
        y_pred = self.global_model.predict(data, verbose=0).flatten()
        return float(mean_squared_error(labels, y_pred))

    def evaluate(self, test_data=None, test_labels=None):
        if test_data is None:
            test_data, test_labels = X_test, y_test
        self.global_model.set_weights(self.global_weights)
        y_pred = self.global_model.predict(test_data, verbose=0).flatten()
        mse = mean_squared_error(test_labels, y_pred)
        return mse


# ============================================================
# 12. Server: FedClust（末层特征聚类 + 簇内采样）
# ============================================================
class ServerFedClust:
    def __init__(self):
        self.global_model = build_model()
        self.global_weights = self.global_model.get_weights()
        self.global_mse_history = []
        self.n_params = sum(w.size for w in self.global_weights)
        self.comm_params = 0
        self.comm_bytes = 0
        self.cluster_models = None
        self.initialized = False

    def _init_clusters(self, clients):
        # 首轮：全部客户端从全局出发本地训练 2 epoch
        all_weights = [c.local_train_for_clustering(self.global_weights, epochs=2)
                       for c in clients]
        self.comm_params += len(clients) * self.n_params
        self.comm_bytes += len(clients) * self.n_params * 4 * 2

        reps = np.stack([c.get_clustering_feature() for c in clients])
        n = len(reps)
        M = np.zeros((n, n))
        for i in range(n):
            for j in range(n):
                M[i, j] = np.linalg.norm(reps[i] - reps[j])
        tau = np.quantile(M[np.triu_indices(n, 1)], FEDCLUST_TAU_QUANTILE)
        hc = AgglomerativeClustering(n_clusters=None,
                                     distance_threshold=max(tau, 1e-6),
                                     linkage='average')
        labels = hc.fit_predict(M)
        for c, lab in zip(clients, labels):
            c.set_cluster_id(int(lab))

        # 【与 FedClust 论文 4.2 对齐】记录每簇的代表特征（末层权重的成员均值），
        # 供新客户端并入时计算与各簇的距离
        self.cluster_features = {}
        for k in set(labels.tolist()):
            members = [i for i in range(n) if labels[i] == k]
            self.cluster_features[k] = reps[members].mean(axis=0)

        # 每簇用成员客户端加权融合的初始簇模型
        self.cluster_models = {}
        for k in set(labels.tolist()):
            members = [i for i in range(n) if labels[i] == k]
            sizes = np.array([clients[i].sample_count() for i in members], dtype=np.float64)
            sizes = sizes / sizes.sum()
            cm = [np.zeros_like(self.global_weights[li]) for li in range(len(self.global_weights))]
            for a, i in zip(sizes, members):
                for li in range(len(self.global_weights)):
                    cm[li] += a * all_weights[i][li]
            self.cluster_models[k] = cm

        self._update_global(clients)
        self.initialized = True
        print(f"FedClust initialized: {len(set(labels.tolist()))} clusters, "
              f"labels={labels.tolist()}")

    def _update_global(self, clients):
        cluster_sizes = collections.defaultdict(int)
        for c in clients:
            cluster_sizes[c.get_cluster_id()] += c.sample_count()
        total = sum(cluster_sizes.values())
        new_w = [np.zeros_like(w) for w in self.global_weights]
        for k in self.cluster_models:
            a = cluster_sizes[k] / total
            for li in range(len(self.global_weights)):
                new_w[li] += a * self.cluster_models[k][li]
        self.global_weights = new_w

    def _allocate_clients(self, clients):
        cluster_ids = sorted(
            self.cluster_models.keys(),
            key=lambda k: -sum(clients[i].sample_count() for i in range(NUM_CLIENTS)
                               if clients[i].get_cluster_id() == k)
        )
        alloc = {}
        remaining = NUM_SELECT_CLIENTS
        for k in cluster_ids:
            cids_k = [i for i in range(NUM_CLIENTS) if clients[i].get_cluster_id() == k]
            a = min(1, remaining, len(cids_k))
            alloc[k] = a
            remaining -= a
            if remaining <= 0:
                break
        if remaining > 0:
            sizes_k = {k: sum(clients[i].sample_count() for i in range(NUM_CLIENTS)
                              if clients[i].get_cluster_id() == k)
                       for k in alloc}
            tot = sum(sizes_k.values()) or 1
            for k in alloc:
                cids_k = [i for i in range(NUM_CLIENTS) if clients[i].get_cluster_id() == k]
                extra = int(remaining * sizes_k[k] / tot)
                alloc[k] = min(len(cids_k), alloc[k] + extra)
        return alloc

    def assign_newcomer(self, client, local_epochs=2):
        """【与 FedClust 论文 Algorithm 2 对齐】新客户端并入：
        先用全局模型本地训练，再计算其末层特征与各簇代表特征的距离，并入最近的簇。
        静态实验中不存在新客户端，故由 FEDCLUST_ENABLE_NEWCOMER=False 关闭；
        该方法作为论文机制的落地接口保留，打开开关即可启用。
        """
        if not self.initialized or not FEDCLUST_ENABLE_NEWCOMER:
            return None
        self.comm_params += self.n_params
        self.comm_bytes += self.n_params * 4 * 2
        client.local_train_for_clustering(self.global_weights, epochs=local_epochs)
        feat = client.get_clustering_feature()
        best_k, best_d = None, np.inf
        for k, cf in self.cluster_features.items():
            d = np.linalg.norm(feat - cf)
            if d < best_d:
                best_k, best_d = k, float(d)
        client.set_cluster_id(int(best_k))
        print(f"FedClust newcomer {client.client_id} -> cluster {best_k} "
              f"(distance {best_d:.4f})")
        return best_k

    def run_round(self, clients):
        if not self.initialized:
            self._init_clusters(clients)
            return
        alloc = self._allocate_clients(clients)
        selected_ids = []
        for k, n_sel in alloc.items():
            cids_k = [i for i in range(NUM_CLIENTS) if clients[i].get_cluster_id() == k]
            if not cids_k or n_sel == 0:
                continue
            chosen = np.random.choice(cids_k, size=min(n_sel, len(cids_k)), replace=False).tolist()
            selected_ids.extend(chosen)
            weights_list, sizes = [], []
            for cid in chosen:
                cw = clients[cid].local_train_from_cluster(self.cluster_models[k])
                weights_list.append(cw)
                sizes.append(clients[cid].sample_count())
            sizes = np.array(sizes, dtype=np.float64)
            sizes = sizes / sizes.sum()
            new_cm = [np.zeros_like(self.global_weights[li]) for li in range(len(self.global_weights))]
            for a, cw in zip(sizes, weights_list):
                for li in range(len(self.global_weights)):
                    new_cm[li] += a * cw[li]
            self.cluster_models[k] = new_cm
            self.comm_params += len(chosen) * self.n_params
            self.comm_bytes += len(chosen) * self.n_params * 4 * 2
        print(f"FedClust selected clients: {sorted(selected_ids)}")
        self._update_global(clients)

    def evaluate(self, test_data=None, test_labels=None):
        if test_data is None:
            test_data, test_labels = X_test, y_test
        self.global_model.set_weights(self.global_weights)
        y_pred = self.global_model.predict(test_data, verbose=0).flatten()
        mse = mean_squared_error(test_labels, y_pred)
        self.global_mse_history.append(mse)
        return mse
# ============================================================
# 13. 单次种子实验
# ============================================================
def run_one_seed(seed):
    set_seeds(seed)
    print(f"\n{'#'*60}\n# SEED {seed}\n{'#'*60}")

    client_data = load_and_partition_housing_non_iid(
        num_clients=NUM_CLIENTS, alpha=0.5, client_ratios=FIXED_RATIOS, val_ratio=0.2
    )

    # 每个算法使用独立的客户端实例集（互不共享状态）
    clients_da = [Client(i, client_data[i]) for i in range(NUM_CLIENTS)]
    clients_gac = [FedGACClient(i, client_data[i]) for i in range(NUM_CLIENTS)]
    clients_aa = [FedAAClient(i, client_data[i]) for i in range(NUM_CLIENTS)]
    clients_clust = [FedClustClient(i, client_data[i]) for i in range(NUM_CLIENTS)]

    s_da = ServerDA()
    s_gac = ServerFedGAC()
    s_aa = ServerFedAA()
    s_clust = ServerFedClust()

    curves = {'FedDAAW': [], 'FedGAC': [], 'FedAA': [], 'FedClust': []}
    stop_round = NUM_GLOBAL_EPOCHS

    for epoch in range(NUM_GLOBAL_EPOCHS):
        print(f"\n{'='*50}\nGlobal Round {epoch+1}/{NUM_GLOBAL_EPOCHS}\n{'='*50}")

        # FedDAAW
        print("\n[FedDAAW Method]")
        sel = s_da.select_clients(clients_da)
        ws = [c.local_train(s_da.global_weights) for c in sel]
        for c in sel:
            c.local_evaluate()
        s_da.aggregate(sel, ws)
        curves['FedDAAW'].append(s_da.evaluate())

        # FedGAC
        print("\n[FedGAC Method]")
        sel = s_gac.select_clients(clients_gac)
        ws = s_gac.local_train_with_ailm(sel)
        s_gac.aggregate(sel, ws)
        curves['FedGAC'].append(s_gac.evaluate())

        # FedAA
        print("\n[FedAA Method]")
        sel = s_aa.select_clients(clients_aa)
        ws = [c.local_train(s_aa.global_weights) for c in sel]
        s_aa.aggregate(sel, ws)
        curves['FedAA'].append(s_aa.evaluate())

        # FedClust
        print("\n[FedClust Method]")
        s_clust.run_round(clients_clust)
        curves['FedClust'].append(s_clust.evaluate())

        print(f"\nRound {epoch+1} Results (Test MSE):")
        for a in ['FedDAAW', 'FedGAC', 'FedAA', 'FedClust']:
            print(f"  {a}: {curves[a][-1]:.4f}")

        # 全局早停：所有方法连续 PATIENCE 轮 MSE 不再下降
        if epoch >= PATIENCE_ROUNDS - 1:
            all_flat = True
            for a in curves:
                h = curves[a]
                if len(h) < PATIENCE_ROUNDS:
                    all_flat = False
                    break
                recent = h[-PATIENCE_ROUNDS:]
                if not all(recent[i] >= recent[i - 1] - IMPROVE_EPS
                           for i in range(1, len(recent))):
                    all_flat = False
                    break
            if all_flat:
                print(f"\nAll methods not improved for {PATIENCE_ROUNDS} rounds, stopping early")
                stop_round = epoch + 1
                break

    results = []
    for algo, srv in [('FedDAAW', s_da), ('FedGAC', s_gac),
                      ('FedAA', s_aa), ('FedClust', s_clust)]:
        curve = curves[algo]
        best = float(min(curve))
        best_r = int(np.argmin(curve)) + 1
        thr = best * 1.05
        r5 = next((i + 1 for i, mse in enumerate(curve) if mse <= thr), len(curve))
        results.append({
            'algorithm': algo, 'seed': seed,
            'final_mse': round(float(curve[-1]), 6),
            'best_mse': round(best, 6),
            'best_round': best_r,
            'stop_round': stop_round,
            'rounds_to_5pct': r5,
            'total_params': srv.comm_params,
            'total_bytes': srv.comm_bytes,
        })
    return results, curves
# ============================================================
# 14. 可视化
# ============================================================
def plot_results(all_curves):
    colors = {'FedDAAW': 'blue', 'FedGAC': 'orange', 'FedAA': 'green', 'FedClust': 'red'}
    markers = {'FedDAAW': 'o', 'FedGAC': 's', 'FedAA': '^', 'FedClust': 'D'}

    plt.figure(figsize=(15, 5))

    # 左：MSE 收敛曲线（多种子 mean ± std）
    plt.subplot(1, 2, 1)
    for algo in colors:
        by_r = collections.defaultdict(list)
        for a, s, r, mse in all_curves:
            if a == algo:
                by_r[r].append(mse)
        rounds = sorted(by_r)
        m = np.array([np.mean(by_r[r]) for r in rounds])
        sd = np.array([np.std(by_r[r]) for r in rounds])
        plt.plot(rounds, m, marker=markers[algo], label=algo, color=colors[algo],
                 linewidth=2, markersize=4)
        plt.fill_between(rounds, m - sd, m + sd, alpha=0.15, color=colors[algo])
    plt.xlabel('Round')
    plt.ylabel('Test MSE')
    plt.title('California Housing Non-IID: MSE (mean ± std over 3 seeds)')
    plt.legend(); plt.grid(alpha=0.5)

    # 右：最终 MSE 柱状图（多种子 mean ± std）
    plt.subplot(1, 2, 2)
    finals = collections.defaultdict(dict)
    for a, s, r, mse in all_curves:
        finals[a].setdefault(s, []).append((r, mse))
    means, stds = [], []
    for a in colors:
        per_seed = [max(v, key=lambda x: x[0])[1] for v in finals[a].values()]
        means.append(np.mean(per_seed))
        stds.append(np.std(per_seed))
    bars = plt.bar(list(colors), means, yerr=stds, capsize=6,
                   color=[colors[a] for a in colors], alpha=0.75, edgecolor='black')
    plt.ylabel('Final Test MSE')
    plt.title('Final MSE (mean ± std)')
    for b, m in zip(bars, means):
        plt.text(b.get_x() + b.get_width() / 2, m + 0.02, f'{m:.4f}',
                 ha='center', va='bottom', fontsize=10)
    plt.grid(alpha=0.6, axis='y')

    plt.tight_layout()
    plt.savefig("results/housing_comparison.png", dpi=150)
    plt.show()
# ============================================================
# 15. 主程序
# ============================================================
def main():
    os.makedirs("results", exist_ok=True)
    all_results, all_curves = [], []

    for seed in SEEDS:
        res, curves = run_one_seed(seed)
        all_results.extend(res)
        for algo, c in curves.items():
            for r, mse in enumerate(c, start=1):
                all_curves.append((algo, seed, r, float(mse)))

    with open("results/result_housing.csv", "w", newline="") as f:
        w = csv.DictWriter(f, fieldnames=list(all_results[0].keys()))
        w.writeheader(); w.writerows(all_results)

    with open("results/convergence_curves_housing.csv", "w", newline="") as f:
        w = csv.writer(f)
        w.writerow(['algorithm', 'seed', 'round', 'test_mse'])
        w.writerows(all_curves)

    print("\n" + "=" * 60)
    print("Aggregated Results (3 seeds, mean ± std)")
    print("=" * 60)
    agg = collections.defaultdict(lambda: collections.defaultdict(list))
    for r in all_results:
        for k in ['final_mse', 'best_mse', 'best_round', 'stop_round',
                  'rounds_to_5pct', 'total_params', 'total_bytes']:
            agg[r['algorithm']][k].append(r[k])
    for algo in ['FedDAAW', 'FedGAC', 'FedAA', 'FedClust']:
        print(f"\n{algo}:")
        for k in agg[algo]:
            v = agg[algo][k]
            print(f"  {k}: {np.mean(v):.4f} ± {np.std(v):.4f}")

    plot_results(all_curves)


if __name__ == "__main__":
    main()
```
